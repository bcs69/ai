```python
from collections import deque

start = ("floor", "monkey", "banana")
goal = ("box", "monkey", "hasbanana")
queue = deque([(start, [start])])
visited = {start}

while queue:
    state, path = queue.popleft()
    if state == goal:
        print("Solution:")
        for step in path:
            print(step)
        break
    moves = []
    if state[0] == "floor":
        moves.append((("box", state[1], state[2]), "Monkey moves to box"))
    if state[0] == "box":
        moves.append((("floor", state[1], state[2]), "Monkey moves down"))
        if state[1] == "monkey" and state[2] == "banana":
            moves.append((("box", "box", "banana"), "Monkey pushes box under banana"))
    if state == ("box", "box", "banana"):
        moves.append((("box", "monkey", "banana"), "Monkey climbs on box"))
    if state == ("box", "monkey", "banana"):
        moves.append((("box", "monkey", "hasbanana"), "Monkey gets banana"))
    for next_state, action in moves:
        if next_state not in visited:
            visited.add(next_state)
            queue.append((next_state, path + [(action, next_state)]))
```
