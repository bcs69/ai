```python
# 10 Marks
import heapq

graph = {
    "V1": {"V2": 2, "V3": 4},
    "V2": {"V4": 3, "V5": 5},
    "V3": {"V5": 2},
    "V4": {"V6": 2},
    "V5": {"V6": 1},
    "V6": {}
}

heuristic = {
    "V1": 5,
    "V2": 4,
    "V3": 3,
    "V4": 2,
    "V5": 1,
    "V6": 0
}

start = "V1"
goal = "V6"

queue = [(heuristic[start], 0, start, [start])]
visited = set()

while queue:
    f, cost, node, path = heapq.heappop(queue)

    if node in visited:
        continue

    visited.add(node)

    if node == goal:
        print("Shortest path:", path)
        print("Cost:", cost)
        break

    for next_node, edge_cost in graph[node].items():
        if next_node not in visited:
            new_cost = cost + edge_cost
            new_f = new_cost + heuristic[next_node]
            heapq.heappush(queue, (new_f, new_cost, next_node, path + [next_node]))
```
